```python
# Adicione a instalação do eSpeak antes de importar pyttsx3
!sudo apt-get update
!sudo apt-get install espeak -y

import pyttsx3
import wikipedia
import webbrowser

# Inicializa o mecanismo de voz
engine = pyttsx3.init()
engine.setProperty('rate', 150)  # Velocidade da fala

def falar(texto):
    print("Assistente:", texto)
    engine.say(texto)
    engine.runAndWait()

def ouvir_comando():
    comando = input("Você (digite seu comando): ").lower()
    return comando

def executar_comando(comando):
    if 'pesquisar por' in comando:
        termo = comando.replace('pesquisar por', '').strip()
        try:
            resultado = wikipedia.summary(termo, sentences=2, lang='pt')
            falar(resultado)
        except wikipedia.exceptions.DisambiguationError:
            falar("Esse termo tem muitos significados. Seja mais específico.")
        except wikipedia.exceptions.PageError:
            falar("Não encontrei nada sobre isso.")
        except:
            falar("Erro ao acessar a Wikipedia.")

    elif 'abrir youtube' in comando:
        webbrowser.open('https://www.youtube.com')
        falar("Abrindo o YouTube.")

    elif 'farmácia mais próxima' in comando:
        webbrowser.open('https://www.google.com/maps/search/farmácia+mais+próxima/')
        falar("Mostrando as farmácias próximas no Google Maps.")

    elif 'sair' in comando or 'encerrar' in comando:
        falar("Encerrando. Até mais!")
        exit()

    else:
        falar("Não entendi o comando. Tente novamente.")

def iniciar_assistente():
    falar("Olá! Sou seu assistente virtual. Digite um comando.")
    while True:
        comando = ouvir_comando()
        executar_comando(comando)

if __name__ == "__main__":
    iniciar_assistente()

```
