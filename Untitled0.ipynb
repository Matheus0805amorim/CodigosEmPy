```python
# Image Similarity Recommendation System
# This script trains a deep learning model to extract image embeddings and recommends similar images based on cosine distance.

import os
import numpy as np
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from sklearn.neighbors import NearestNeighbors
import pickle

# 1. Load images from directory and compute embeddings
def load_and_embed(image_dir, model, target_size=(224,224)):
    embeddings = []
    file_paths = []
    for fname in os.listdir(image_dir):
        fpath = os.path.join(image_dir, fname)
        if not fname.lower().endswith(('.jpg', '.png', '.jpeg')):
            continue
        img = image.load_img(fpath, target_size=target_size)
        x = image.img_to_array(img)
        x = np.expand_dims(x, axis=0)
        x = preprocess_input(x)
        feat = model.predict(x)
        feat_norm = feat / np.linalg.norm(feat)
        embeddings.append(feat_norm.flatten())
        file_paths.append(fpath)
    return np.array(embeddings), file_paths

# 2. Build recommendation index using NearestNeighbors

def build_index(embeddings, n_neighbors=5):
    nn = NearestNeighbors(n_neighbors=n_neighbors, metric='cosine')
    nn.fit(embeddings)
    return nn

# 3. Query function to recommend similar images
def recommend(query_path, model, index, file_paths, top_n=5):
    img = image.load_img(query_path, target_size=(224,224))
    x = image.img_to_array(img)
    x = np.expand_dims(x, axis=0)
    x = preprocess_input(x)
    feat = model.predict(x)
    feat_norm = feat / np.linalg.norm(feat)
    dists, idxs = index.kneighbors(feat_norm.flatten().reshape(1, -1), n_neighbors=top_n)
    results = [(file_paths[i], dists[0][j]) for j, i in enumerate(idxs[0])]
    return results

if __name__ == '__main__':
    IMAGE_DIR = 'dataset/images'
    EMBEDDING_MODEL = MobileNetV2(weights='imagenet', include_top=False, pooling='avg')

    # Compute embeddings for dataset
    embeddings, paths = load_and_embed(IMAGE_DIR, EMBEDDING_MODEL)

    # Build index
    nn_index = build_index(embeddings, n_neighbors=6)

    # Save model and index
    with open('nn_index.pkl', 'wb') as f:
        pickle.dump({'index': nn_index, 'paths': paths}, f)

    # Example recommendation
    query_img = 'dataset/query/shoe.jpg'
    recs = recommend(query_img, EMBEDDING_MODEL, nn_index, paths, top_n=6)
    print("Recommendations for {}:".format(query_img))
    for path, dist in recs:
        print(f"{path} (distance: {dist:.4f})")
```
